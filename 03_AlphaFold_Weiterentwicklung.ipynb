pLDDT-Farben (wie in der AlphaFold DB): sehr hoch (>90) blau, hoch (70–90) cyan, niedrig (50–70) gelb, sehr niedrig (<50) orange.


```python
# In Colab beim ersten Lauf:
# %pip install -q py3Dmol biopython

import io
import json
import statistics
import urllib.request

import py3Dmol
from Bio.Align import PairwiseAligner
from Bio.PDB import PDBParser, Superimposer
from Bio.PDB.Polypeptide import is_aa, protein_letters_3to1

USER_AGENT = "rosary-mediteam-notebook-03"

EXAMPLES = [
    ("P00367", "GLUD1", "Glutamat-Dehydrogenase 1, mitochondrial"),
    ("P02452", "COL1A1", "Kollagen alpha-1(I)-Kette"),
    ("P04406", "GAPDH", "Glycerinaldehyd-3-phosphat-Dehydrogenase"),
]

# H²-M: selbst gesetzte Schwelle, kein Literaturwert.
H2M_RMSD_THRESHOLD_A = 2.0
PLDDT_HIGH = 90.0


def _fetch(url: str, accept: str | None = None, timeout: int = 60) -> bytes:
    headers = {"User-Agent": USER_AGENT}
    if accept:
        headers["Accept"] = accept
    req = urllib.request.Request(url, headers=headers)
    with urllib.request.urlopen(req, timeout=timeout) as resp:
        return resp.read()


def alphafold_model(uniprot_id: str) -> dict:
    """Aktuelle Modell-URL über die AlphaFold-API, Fallback auf das Dateimuster."""
    uid = uniprot_id.strip().upper()
    api = f"https://alphafold.ebi.ac.uk/api/prediction/{uid}"
    recs = json.loads(_fetch(api, accept="application/json", timeout=30))
    rec = recs[0] if isinstance(recs, list) else recs
    version = rec.get("latestVersion") or rec.get("modelVersion") or 4
    pdb_url = rec.get("pdbUrl") or (
        f"https://alphafold.ebi.ac.uk/files/AF-{uid}-F1-model_v{version}.pdb"
    )
    return {
        "uniprot": uid,
        "version": version,
        "pdb_url": pdb_url,
        "entry": rec.get("entryId") or f"AF-{uid}",
        "gene": rec.get("gene") or rec.get("uniprotGene") or "",
        "name": rec.get("uniprotDescription") or "",
        "organism": rec.get("organismScientificName") or rec.get("organism") or "",
        "global_plddt": rec.get("globalMetricValue"),
        "created": rec.get("modelCreatedDate"),
        "raw": rec,
    }


def load_pdb_text(url: str) -> str:
    return _fetch(url, timeout=60).decode("utf-8", errors="replace")


def _colorfunc_js() -> str:
    return (
        "function(atom){"
        "if(atom.b>90) return '#0053D6';"
        "if(atom.b>70) return '#65CBF3';"
        "if(atom.b>50) return '#FFDB13';"
        "return '#FF7D45';}"
    )


def _maybe_show(view) -> None:
    """3D-Ansicht in Jupyter/Colab zeigen; im reinen Lauf nur vermerken."""
    try:
        from IPython import get_ipython
        from IPython.display import display

        if get_ipython() is not None:
            display(view)
            return
    except Exception as exc:  # noqa: BLE001
        print(f"*anzeige: 3D-Ansicht nicht gerendert ({type(exc).__name__}). In Colab erscheint sie hier.")
        return
    print("*anzeige: keine interaktive Jupyter-Sitzung; 3D-Ansicht in Colab sichtbar.")


def view_human_alphafold(uniprot_id: str, width=640, height=480):
    meta = alphafold_model(uniprot_id)
    pdb = load_pdb_text(meta["pdb_url"])
    view = py3Dmol.view(width=width, height=height)
    view.addModel(pdb, "pdb")
    view.setStyle({"cartoon": {"colorfunc": _colorfunc_js()}})
    view.zoomTo()
    print(
        f"{meta['uniprot']}  {meta['gene']}  {meta['organism']}\n"
        f"{meta['name']}\n"
        f"AlphaFold-DB-Version {meta['version']}  global pLDDT {meta['global_plddt']}\n"
        f"{meta['pdb_url']}\n"
        f"Eintrag https://alphafold.ebi.ac.uk/entry/{meta['uniprot']}\n"
        "Färbung: pLDDT (B-Faktor). Struktur-Vorhersage (AlphaFold DB). "
        "Mögliche Wirkungen sind Gegenstand unserer F&E-Hypothesen (H1/H²)."
    )
    _maybe_show(view)
    return meta, pdb
```

## Strukturvergleich H²-M

Vorab festgelegte Aussage (selbst gesetzte Schwelle, kein Literaturwert): Für menschliche Referenzproteine mit experimenteller Struktur sollen Bereiche mit pLDDT > 90 nach Überlagerung höchstens **2,0 Å Cα-RMSD** zur experimentellen Struktur aufweisen.

- GAPDH (P04406) gegen PDB [1U8F](https://www.rcsb.org/structure/1U8F) (human, 1,75 Å, Röntgen)
- GLUD1 (P00367) gegen PDB [1L1F](https://www.rcsb.org/structure/1L1F) (human, apo, 2,7 Å, Röntgen)

Methode: Sequenzalignment (Biopython `PairwiseAligner`), Überlagerung der Cα-Atome (`Bio.PDB.Superimposer`). Alle Proteinketten der Kristallstruktur werden einzeln verglichen; berichtet werden jede Kette und die Kette mit dem niedrigsten RMSD bei pLDDT > 90. Ein einzelner Lauf gilt nicht als unabhängige Bestätigung (`*lauf`).

Struktur-Vorhersage (AlphaFold DB). Mögliche Wirkungen sind Gegenstand unserer F&E-Hypothesen (H1/H²).


```python
def parse_structure(pdb_text: str, name: str):
    return PDBParser(QUIET=True).get_structure(name, io.StringIO(pdb_text))


def aa_ca(chain):
    records = []
    for residue in chain:
        if not is_aa(residue, standard=True) or "CA" not in residue:
            continue
        letter = protein_letters_3to1.get(residue.get_resname())
        if not letter:
            continue
        ca = residue["CA"]
        records.append(
            {
                "resseq": residue.id[1],
                "aa": letter,
                "ca": ca,
                "b": float(ca.get_bfactor()),
            }
        )
    return records


def protein_chains(structure):
    chains = []
    for model in structure:
        for chain in model:
            recs = aa_ca(chain)
            if recs:
                chains.append((chain.id, recs))
    return chains


def align_pairs(af_recs, exp_recs, plddt_min=PLDDT_HIGH):
    aligner = PairwiseAligner()
    aligner.mode = "global"
    aligner.match_score = 2
    aligner.mismatch_score = -1
    aligner.open_gap_score = -10
    aligner.extend_gap_score = -0.5
    af_seq = "".join(r["aa"] for r in af_recs)
    exp_seq = "".join(r["aa"] for r in exp_recs)
    alignment = aligner.align(af_seq, exp_seq)[0]
    pairs_all, pairs_hi = [], []
    for (a0, a1), (e0, e1) in zip(*alignment.aligned):
        for i, j in zip(range(a0, a1), range(e0, e1)):
            pair = (af_recs[i], exp_recs[j])
            pairs_all.append(pair)
            if af_recs[i]["b"] > plddt_min:
                pairs_hi.append(pair)
    return alignment, pairs_all, pairs_hi


def ca_rmsd(pairs):
    if len(pairs) < 3:
        return None
    ref = [p[1]["ca"] for p in pairs]
    mob = [p[0]["ca"] for p in pairs]
    sup = Superimposer()
    sup.set_atoms(ref, mob)
    return float(sup.rms)


def compare_h2m(uniprot_id: str, pdb_id: str, label: str) -> list[dict]:
    meta = alphafold_model(uniprot_id)
    af_text = load_pdb_text(meta["pdb_url"])
    exp_text = load_pdb_text(f"https://files.rcsb.org/download/{pdb_id.upper()}.pdb")
    af_chains = protein_chains(parse_structure(af_text, f"AF-{uniprot_id}"))
    exp_chains = protein_chains(parse_structure(exp_text, pdb_id))
    af_id, af_recs = af_chains[0]
    print("=" * 70)
    print(f"H²-M  {label}")
    print(
        f"AF {meta['entry']} v{meta['version']}  Kette {af_id}: "
        f"{len(af_recs)} Reste, mittlerer pLDDT "
        f"{statistics.mean(r['b'] for r in af_recs):.2f}, "
        f"pLDDT>{PLDDT_HIGH:.0f}: {sum(1 for r in af_recs if r['b'] > PLDDT_HIGH)}"
    )
    print(
        f"EXP PDB {pdb_id}  {len(exp_chains)} Proteinketten. "
        f"Schwelle ≤ {H2M_RMSD_THRESHOLD_A:.1f} Å Cα-RMSD (selbst gesetzt)."
    )
    rows = []
    for ch_id, exp_recs in exp_chains:
        _, pairs_all, pairs_hi = align_pairs(af_recs, exp_recs)
        ident = sum(a["aa"] == e["aa"] for a, e in pairs_all) / max(len(pairs_all), 1)
        rms_all = ca_rmsd(pairs_all)
        rms_hi = ca_rmsd(pairs_hi)
        row = {
            "label": label,
            "uniprot": uniprot_id,
            "pdb": pdb_id,
            "af_version": meta["version"],
            "exp_chain": ch_id,
            "n_all": len(pairs_all),
            "rmsd_all": rms_all,
            "n_hi": len(pairs_hi),
            "rmsd_hi": rms_hi,
            "identity": ident,
            "threshold": H2M_RMSD_THRESHOLD_A,
            "within_threshold": (
                rms_hi is not None and rms_hi <= H2M_RMSD_THRESHOLD_A
            ),
        }
        rows.append(row)
        flag = "innerhalb der Schwelle" if row["within_threshold"] else "über der Schwelle"
        print(
            f"  Kette {ch_id}: aligned={row['n_all']}  Identität={ident:.3f}  "
            f"Cα-RMSD alle={rms_all:.3f} Å  "
            f"Cα-RMSD pLDDT>{PLDDT_HIGH:.0f} (n={row['n_hi']})={rms_hi:.3f} Å  → {flag}"
        )
    best = min(rows, key=lambda r: (r["rmsd_hi"] is None, r["rmsd_hi"] or 99.0))
    print(
        f"  Beste Kette {best['exp_chain']}: "
        f"{best['rmsd_hi']:.3f} Å (pLDDT>{PLDDT_HIGH:.0f}, n={best['n_hi']}); "
        f"{'≤' if best['within_threshold'] else '>'} {H2M_RMSD_THRESHOLD_A:.1f} Å."
    )
    return rows


h2m_rows = []
h2m_rows.extend(
    compare_h2m("P04406", "1U8F", "GAPDH (P04406) gegen 1U8F")
)
h2m_rows.extend(
    compare_h2m("P00367", "1L1F", "GLUD1 (P00367) gegen 1L1F")
)

print("\nZusammenfassung H²-M (Cα-RMSD, Biopython Superimposer)")
print(
    f"{'Vergleich':<32} {'Kette':<6} {'n_all':>6} {'RMSD_all':>9} "
    f"{'n_>90':>6} {'RMSD_>90':>9} {'≤2.0 Å':>7}"
)
for row in h2m_rows:
    print(
        f"{row['label']:<32} {row['exp_chain']:<6} {row['n_all']:6d} "
        f"{row['rmsd_all']:8.3f} Å {row['n_hi']:6d} "
        f"{row['rmsd_hi']:8.3f} Å {'ja' if row['within_threshold'] else 'nein':>7}"
    )
```

## * Grenzen

Ein `*`-Fenster klassifiziert eine Einschränkung. Es sperrt den Lauf nicht. Ein einzelner deterministischer Lauf gilt nicht als unabhängige Bestätigung.

| Fenster | Klasse | Text |
|---|---|---|
| `*modell` | Datenbank | Die Struktur kommt aus der AlphaFold DB (EMBL-EBI). Sie ist eine Vorhersage, kein Kristall- oder NMR-Assay und kein selbst gerechneter AlphaFold-Lauf. |
| `*version` | API | Die Dateiversion (am 18.11.2026: v6, nicht fest v4) wird über `https://alphafold.ebi.ac.uk/api/prediction/<ID>` ermittelt. |
| `*plddt` | Skala | Die Färbung und der Filter pLDDT > 90 folgen dem B-Faktor der AFDB-PDB-Datei. Das ist eine Konfidenz der Vorhersage, kein Assay-Messwert einer Anwendung. |
| `*schwelle` | Annahme | Die Schwelle ≤ 2,0 Å Cα-RMSD in H²-M ist unsere eigene, vorab gesetzte Festlegung und kein Literaturwert. |
| `*cmp` | Vergleich | Der RMSD gilt für Cα-Atome nach Sequenzalignment. 1U8F ist ein Tetramer, 1L1F ein Hexamer (apo, 2,7 Å); die AFDB liefert eine monomere Vorhersage. Liganden- oder Domänenbewegungen können den Wert beeinflussen. |
| `*lauf` | Wiederholung | Dies ist ein erster Lauf. Als *decision-capable* gilt die Kette erst, wenn ein zweiter, unabhängiger Lauf oder ein anderes Format sie stützt. |
| `*h1` | Umfang | Struktur-Vorhersage (AlphaFold DB). Mögliche Wirkungen sind Gegenstand unserer F&E-Hypothesen (H1/H²). |
